```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

np.random.seed(42)

categories = ["Electronics", "Clothing", "Furniture", "Books"]

data = pd.DataFrame({
    "Product ID": range(1, 101),
    "Product Category": np.repeat(categories, 25),
    "Product Name": ["Product " + str(i) for i in range(1, 101)],
    "Brand": np.random.choice(["Samsung", "Nike", "IKEA", "Sony"], 100),
    "Price": np.random.randint(100, 10000, 100),
    "Customer Rating": np.round(np.random.uniform(2.5, 5, 100), 1)
})

data.to_csv("products.csv", index=False)

data = pd.read_csv("products.csv")

data.fillna(data.median(numeric_only=True), inplace=True)

sns.set_theme(style="whitegrid")

fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 5))

sns.boxplot(data=data, x="Product Category", y="Price",
            palette=["red", "blue", "green", "orange"],
            ax=axes[0], hue="Product Category", legend=False)

axes[0].set_title("Product Price Distribution (Box Plot)")
axes[0].set_xlabel("Product Category")
axes[0].set_ylabel("Price")

sns.violinplot(data=data, x="Product Category", y="Price",
               palette=["red", "blue", "green", "orange"],
               ax=axes[1], hue="Product Category", legend=False)

axes[1].set_title("Product Price Distribution (Violin Plot)")
axes[1].set_xlabel("Product Category")
axes[1].set_ylabel("Price")

fig.suptitle("Product Price Analysis")

plt.tight_layout()
plt.savefig("product_analysis.png")
plt.show()

print(data["Price"].describe())

print(data.groupby("Product Category")["Price"].median())
```
